# N-GRAMS

```python
a = "i am afnan khan i am a student i am a good boy a afnan khan"
```

```python
from nltk.tokenize import word_tokenize
```

```python
w = word_tokenize(a)
```

```python
from nltk import ngrams
from nltk.collocations import BigramCollocationFinder, TrigramCollocationFinder
```

```python
b = BigramCollocationFinder.from_words(w)    #BigramCollocationFinder: Do shabdon (bigrams) ki common jodi dhoondta hai, jaise "New York."
t = TrigramCollocationFinder.from_words(w)      #TrigramCollocationFinder: Teen shabdon (trigrams) ki common jodi dhoondta hai, jaise "I am happy."
n = ngrams(w, 4)                     #   Kisi bhi length ke shabdon ke combinations banata hai, jaise "I am" (bigram) ya "I am happy" (trigram) or suggesest krta he ki aage isko use kre
```

```python
b.ngram_fd
t.ngram_fd
```

```python
b.ngram_fd.keys()     # ngram_fd: N-grams ki ginti karta hai, jaise "I am" kitni baar aaya.

t.ngram_fd.keys()      # ngram_fd.keys(): Yeh unique n-grams ko dikhata hai, jaise "I am", "am afnan".
```

```python
for i in n:
    print(i)
```
